### Давайте импортируем все разом

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Сегодня понадобится много всяких моделек

```python
from sklearn.neighbors import KNeighborsRegressor as kNN
from sklearn.linear_model import LinearRegression as LinReg
from sklearn.linear_model import ElasticNet, ElasticNetCV
from sklearn.linear_model import Ridge, RidgeCV
from sklearn.linear_model import Lasso, LassoCV
```

Не забываем про подготовку данных

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

Для удобства можно написать подобную функцию

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

def train_test_model(model, X_train, y_train, X_test, y_test):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)

    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    return pd.DataFrame({'MAE': [mae], 'R2': [r2]})
```

### Задача 1. Диабет

Целевой переменной в этой задаче является количественная мера прогрессии диабета за год. Сравните точность (можно любыми метриками для регрессии, но R2 — классика) для обычной линейной регрессии, а так же с L2- и L1-регуляризацией. Постройте графики зависимости точности от силы регуляризации с и без стандартизацией признаков.

```python
from sklearn.datasets import load_diabetes
data = load_diabetes(as_frame=True, scaled=False)
data
```

```python
X = data.data
y = data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

```python
scaler = StandardScaler()
scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
models = [LinReg(), Ridge(), Lasso()]
features = [(X_train, X_test), (X_train_scaled, X_test_scaled)]

scores = []
for model in models:
  for X_train, X_test in features:
    result = train_test_model(model, X_train, y_train, X_test, y_test)
    scores.append(result.loc[0, 'R2'])

scores
```

```python
pd.DataFrame({
    'model': ['LR']*2 + ['Ridge']*2 + ['LASSO']*2,
    'features': ['not_scaled', 'scaled']*3,
    'R2': scores
})
# просто лин трансформация - для лин регрессии не нужно шкалировать, сама справляется. но это удобно для интерпретации
```

```python
model = LinReg()
res = train_test_model(model, X_train, y_train, X_test, y_test)
res.loc[0,'R2']

#model.fit(X_train, y_train)
model.fit(X_train_scaled, y_train)
model.coef_
pd.Series(dict(zip(X_train.columns, model.coef_))).abs().sort_values().plot.bar()

```

```python
model = Ridge()
model.fit(X_train, y_train)
#model.fit(X_train_scaled, y_train)
model.coef_
pd.Series(dict(zip(X_train.columns, model.coef_))).abs().sort_values().plot.bar()
```

### Задача 2. Цены домов в Калифорнии

Целевая переменная — медианная стоимость дома в округах Калифорнии, выраженная в сотнях тысяч долларов. Попробуйте подобрать гиперпараметры (число соседей и меру расстояния) для регрессора kNN. Сравните качество с LinReg, RidgeCV, LassoCV и ElasticNetCV. Постройте столбчатую диаграмму для визуальной оценки.

```python
from sklearn.datasets import fetch_california_housing
data = fetch_california_housing(as_frame=True)
X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X
```

```python
model = kNN()
train_test_model(model, X_train, y_train, X_test, y_test)
# получше, предполагает нелинейность
```

```python
model = LinReg()
train_test_model(model, X_train, y_train, X_test, y_test)
```

```python
model = RidgeCV()
train_test_model(model, X_train, y_train, X_test, y_test)
```

```python
model = LassoCV()
train_test_model(model, X_train, y_train, X_test, y_test)
```

```python
model = ElasticNetCV()
train_test_model(model, X_train, y_train, X_test, y_test)
```
