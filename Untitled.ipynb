```python
from pathlib import Path
from html.parser import HTMLParser
import re

INPUT_DIR = Path(r"C:\Users\Арманта\Downloads\Telegram Desktop\ChatExport_2026-11-25 (1)")      # папка с исходными .html файлами
OUTPUT_DIR = Path(r"C:\Users\Арманта\Downloads\Telegram Desktop\ChatExport_2026-11-25 (1)\cleaned_txt")    # куда сохранить очищенные .txt файлы
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


class TextExtractor(HTMLParser):
    def __init__(self):
        super().__init__()
        self.skip_tags = {"style", "script", "head"}
        self.block_tags = {"div", "p", "br", "tr", "li", "h1", "h2", "h3", "h4"}
        self._skip_depth = 0
        self.chunks = []

    def handle_starttag(self, tag, attrs):
        if tag in self.skip_tags:
            self._skip_depth += 1
        elif tag in self.block_tags and self._skip_depth == 0:
            self.chunks.append("\n")

    def handle_endtag(self, tag):
        if tag in self.skip_tags and self._skip_depth > 0:
            self._skip_depth -= 1

    def handle_data(self, data):
        if self._skip_depth == 0:
            text = data.strip()
            if text:
                self.chunks.append(text + " ")


def clean_html_file(src: Path, dst: Path):
    raw = src.read_text(encoding="utf-8", errors="ignore")
    parser = TextExtractor()
    parser.feed(raw)
    cleaned = "".join(parser.chunks)
    # Убираем лишние пустые строки и пробелы
    cleaned = re.sub(r"[ \t]+", " ", cleaned)
    cleaned = re.sub(r"\n\s*\n+", "\n\n", cleaned).strip()
    dst.write_text(cleaned, encoding="utf-8")


for html_file in sorted(INPUT_DIR.glob("*.html")):
    out_file = OUTPUT_DIR / f"{html_file.stem}.txt"
    clean_html_file(html_file, out_file)

print("Готово! Обработано файлов:", len(list(OUTPUT_DIR.glob("*.txt"))))
```
